## BeautifulSoup Web Scraping

```python
# Importing required packages

import requests
from bs4 import BeautifulSoup
import pandas as pd
```

### Source 1 - Parks in Seoul

```python
# Storing webpage URL
parks_url = "https://en.wikipedia.org/wiki/List_of_parks_in_Seoul"
```

```python
# Sending Requests

headers = {"User-Agent": "Mozilla/5.0"}
response = requests.get(parks_url, headers=headers)
print(response)
```

```python
# Parsing the HTML and finding the table

soup = BeautifulSoup(response.text, "html.parser")
table = soup.find("table", {"class": "wikitable"})
rows = table.find_all("tr")
print(len(rows))
```

```python
# Checking the structure of data
for row in rows[:3]:
    columns = row.find_all("td")
    for column in columns:
        print(column.get_text(strip=True))
    print("----------")
```

```python
# Looping through the first 5 rows and building our list
# Column order in this table is: Parks, Korean name, Location, Note
park_data = []
for row in rows[1:6]:
    columns = row.find_all("td")
    if len(columns) >= 4:
        attraction = columns[0].get_text(strip=True)
        area = columns[3].get_text(strip=True)
        park_data.append([attraction, "Park", area])

park_data
```

### Source 2 - Museums in Seoul


```python
# Storing webpage URL

museums_url = "https://en.wikipedia.org/wiki/List_of_museums_in_Seoul"
```

```python
# Sending Requests

response = requests.get(museums_url, headers=headers)
print(response)
```

```python
# Parsing the HTML and finding the table

soup = BeautifulSoup(response.text, "html.parser")
table = soup.find("table", {"class": "wikitable"})
rows = table.find_all("tr")
print(len(rows))
```

```python
# Checking the structure of data

for row in rows[:3]:
    columns = row.find_all("td")
    for column in columns:
        print(column.get_text(strip=True))
    print("----------")
```

```python
# Looping through the first 5 rows and building our list
# Column order in this table is: Museum, Korean name, Location, Note

museum_data = []
for row in rows[1:6]:
    columns = row.find_all("td")
    if len(columns) >= 3:
        attraction = columns[0].get_text(strip=True)
        area = columns[2].get_text(strip=True)
        museum_data.append([attraction, "Museum", area])

museum_data
```

### Combining both sources into one CSV

```python
# Combining the two lists into one list

attraction_data = park_data + museum_data
```

```python
# Converting list into dataframe

seoul_attractions_df = pd.DataFrame(
    attraction_data,
    columns=["Attraction", "Category", "Area"]
)

print(seoul_attractions_df)
```

```python
# Exporting CSV file

seoul_attractions_df.to_csv("seoul_attractions.csv", index=False)
```
