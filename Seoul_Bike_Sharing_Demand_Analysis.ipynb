## 7. Feature Engineering
Creating new helper columns (time-of-day and weather buckets) to make grouped analysis easier.

```python
# Create a new Time_Category column, grouping each hour into Night / Morning / Afternoon / Evening
df['Time_Category']=np.select([((df['Hour']>=22) & (df['Hour']<=23 ))| (df['Hour']>=0 )& (df['Hour']<=4),(df['Hour']>=5) & (df['Hour']<=11),(df['Hour']>=12) & (df['Hour']<=16),(df['Hour']>=17) & (df['Hour']<=21)],["Night","Morning","Afternoon","Evening"], default= "invalid")
```

```python
# Preview the new Time_Category column
df['Time_Category']
```

```python
# Create a new Weather Category column based on Rainfall amount (No Rain / Light Rain / Heavy Rain)
df['Weather Category']=np.select([df['Rainfall(mm)']==0,(df['Rainfall(mm)']>0)&(df['Rainfall(mm)']<=18),(df['Rainfall(mm)']>18 )&(df['Rainfall(mm)']<=35)],["No Rain","Light Rain","Heavy Rain"], default="invalid")
```

## 8. Key Insights
Answering specific questions about rental demand using groupby and aggregate functions.

```python
# Total number of bikes rented across the entire dataset
df['Rented Bike Count'].sum()
```

```python
# Average number of bikes rented per hour, rounded to 2 decimal places
round(df['Rented Bike Count'].mean(),2)
```

```python
# Average bike rentals for each hour of the day
round(df.groupby('Hour')['Rented Bike Count'].mean(),2)
```

```python
# Highest average hourly rental count across all hours
round(df.groupby('Hour')['Rented Bike Count'].mean().max(),2)
```

```python
# Hour of the day with the highest average rental count
df.groupby('Hour')['Rented Bike Count'].mean().idxmax()
```

```python
# Total bikes rented in each season
df.groupby('Seasons')['Rented Bike Count'].sum()
```

```python
# Season with the highest total rentals
df.groupby('Seasons')['Rented Bike Count'].sum().idxmax()
```

```python
# Total bikes rented on holidays vs non-holidays
df.groupby('Holiday')['Rented Bike Count'].sum()
```

```python
# Category (Holiday / No Holiday) with the higher total rentals
df.groupby('Holiday')['Rented Bike Count'].sum().idxmax()
```

```python
# Total bikes rented under each weather category
df.groupby('Weather Category')['Rented Bike Count'].sum()
```

```python
# Weather category with the highest total rentals
df.groupby('Weather Category')['Rented Bike Count'].sum().idxmax()
```

```python
# Weather category with the lowest total rentals
df.groupby('Weather Category')['Rented Bike Count'].sum().idxmin()
```
